```python
import pandas as pd
```

```python
df = pd.read_csv(r'C:\Users\mager\Downloads\archive (1)\branches.csv' , parse_dates = ['opened_date'])
```

```python
print(df.head)
```

```python
print(df.dtypes)
```

```python
df['ifsc_code'].str.match(r'^[A-Z]{4}0[A-Z0-9]{6}$')
```

```python
df['months'] = df['opened_date'].dt.month_name()
```

```python
df['days'] = df['opened_date'].dt.day
```

```python
df = df[df['opened_date'] < pd.to_datetime('today')]
```

```python
df['branch_name'] = df['branch_name'].astype(str).str.strip().str.lower()
```

```python
df['city'] = df['city'].astype(str).str.strip().str.lower()
```

```python
df['state'] = df['state'].astype(str).str.strip().str.lower()
```

```python
df.to_csv(
    "branches.csv",
    index=False
)
```
