```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\mager\Downloads\archive (1)\customers.csv", parse_dates = ["date_of_birth" , "join_date"])
```

```python
print(df.head())
```

```python
print(df.info())
```

```python
df['customer_id'] = df['customer_id'].abs().astype(int)
```

```python
df['name'] = df['name'].str.strip()
```

```python
df['name'] = df['name'].str.title()
```

```python
print(df['gender'].unique())
```

```python
df['gender'] = df['gender'].str.strip()
```

```python
today = pd.to_datetime('today')
```

```python
future_dates = df[(df['date_of_birth'] > today) | (df['join_date'] > today)]
print(future_dates)
```

```python
df['city'] = df['city'].str.strip()
```

```python
df['city'] = df['city'].astype(str).str.capitalize()
```

```python
df['state'] = df['state'].str.strip()
```

```python
today = pd.to_datetime('today')
```

```python
df['age_years'] = ((today - df['date_of_birth']).dt.days /365.25 ).astype(int)
print(df['age_years'])
```

```python
df['joining_year'] = df['join_date'].dt.year
```

```python
df['joining_month'] =df['join_date'].dt.month
```

```python
def income_category(income):
    if income < 50000:
        return 'Low'
    elif income < 100000:
        return 'Medium'
    else:
        return 'High'

df['annual_category'] = df['annual_income'].apply(income_category)
print(df['annual_category'])
```

```python
df.to_csv(
    "customers.csv",
    index=False
)
```
